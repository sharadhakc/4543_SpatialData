# 🧼 Cleaning Dirty Data (Missing Values & Type Fixes)

## 🔹 LEARNING GOALS:
- Detect and count missing values (`NaN`)
- Fill or drop missing data
- Convert column data types safely
- Understand the difference between `NaN`, `None`, `""`, and type mismatches


### 🧪 1. Load a Messy Dataset

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", None],
    "Age": ["25", "thirty", 35, np.nan, "40"],
    "Signup Date": ["2022-01-01", "not a date", "2022/03/01", None, "April 5, 2022"],
    "Score": [95.5, None, 88.0, 92.5, ""],
    "Notes": ["", None, "active", None, ""]
}

df = pd.DataFrame(data)
df
```

### 🧯 2. Detecting Missing or Broken Values

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

### 🧹 3. Cleaning Strategy Options

```python
df.fillna({
    "Name": "Unknown",
    "Age": -1,
    "Signup Date": "1970-01-01",
    "Score": 0.0,
    "Notes": "N/A"
})
```

```python
df.dropna()
```

### 🧬 4. Data Type Fixes

```python
df.dtypes
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Score"] = pd.to_numeric(df["Score"], errors="coerce")
df["Signup Date"] = pd.to_datetime(df["Signup Date"], errors="coerce")

df.dtypes
```

### 🩹 5. Impute (Fill In) Fixed Missing Values

```python
df["Age"].fillna(df["Age"].median())
df["Score"].fillna(df["Score"].mean())
```

```python
df["Signup Date"].fillna(df["Signup Date"].min())
```

```python
df["Signup Date"].fillna("No Notes yet")
```

```python
df["Name"].fillna("Unknown")
```

### 🤓 6. Cleaned Data Review

```python
df.info()
```

```python
df.describe(include="all")
```

### 🧠 Mini-Challenge

> 🗂 Load `"data/survey.csv"` and:
- Identify which columns have missing values
- Use `.isnull().sum()` to get a null report
- Use a mix of `.fillna()`, `.dropna()`, and `pd.to_numeric()` or `pd.to_datetime()` to clean it
- Print a summary with `.info()` and `.describe()`

```python

df = pd.read_csv("../../data/survey.csv")
df
```

```python
print("Missing values per column:")
print(df.isnull().sum())
```

```python
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["gpa"] = pd.to_numeric(df["gpa"], errors="coerce")
df["study_hours_per_week"] = pd.to_numeric(df["study_hours_per_week"], errors="coerce")
df["satisfaction_1_5"] = pd.to_numeric(df["satisfaction_1_5"], errors="coerce")
df["commute_minutes"] = pd.to_numeric(df["commute_minutes"], errors="coerce")
df["attendance_rate"] = pd.to_numeric(df["attendance_rate"].str.replace("%", ""), errors="coerce")
df["survey_date"] = pd.to_datetime(df["survey_date"], errors="coerce")
```

```python
df["age"] = df["age"].fillna(df["age"].median())
df["gpa"] = df["gpa"].fillna(df["gpa"].median())
df["study_hours_per_week"] = df["study_hours_per_week"].fillna(df["study_hours_per_week"].median())
df["commute_minutes"] = df["commute_minutes"].fillna(df["commute_minutes"].median())
df["gender"] = df["gender"].fillna("Unknown")
df["major"] = df["major"].fillna("Unknown")
df = df.dropna(subset=["survey_date"])
```

```python
print(df.info())
print(df.describe())
```
