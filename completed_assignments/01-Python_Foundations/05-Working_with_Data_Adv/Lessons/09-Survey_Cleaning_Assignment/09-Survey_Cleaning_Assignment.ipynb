```python
import pandas as pd
df = pd.read_csv("survey_dirty.csv")
print(df.head())
df.info()
```

```python
df['name'] = df['name'].str.strip()
df['email'] = df['email'].str.strip()
df = df.replace(r'^\s*$', pd.NA, regex=True)
```

```python
print(df.isnull().sum())

df['age'] = df['age'].fillna(df['age'].median())
df['weight_kg'] = df['weight_kg'].fillna(df['weight_kg'].median())
df['height_cm'] = df['height_cm'].fillna(df['height_cm'].median())
df = df.dropna(subset=['name', 'email'])
```

```python
df['gender'] = df['gender'].str.strip().str.title()
df['gender'] = df['gender'].replace({
    'M': 'Male', 'F': 'Female',
    'Male ': 'Male', 'female': 'Female'
})

df['exercise_freq'] = df['exercise_freq'].str.strip().str.lower()
df['exercise_freq'] = df['exercise_freq'].replace({
    'every day': 'Daily',
    'daily': 'Daily',
    'once a week': 'Weekly',
    'weekly': 'Weekly',
    'never': 'Never',
    'rarely': 'Rarely'
})
```

```python
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['height_cm'] = df['height_cm'].str.replace('cm', '').pipe(pd.to_numeric, errors='coerce')
df['weight_kg'] = df['weight_kg'].str.replace('kg', '').pipe(pd.to_numeric, errors='coerce')
df['satisfaction'] = pd.to_numeric(df['satisfaction'], errors='coerce')
df['submitted_at'] = pd.to_datetime(df['submitted_at'], errors='coerce')
```

```python
email_pattern = r'^[\w\.-]+@[\w\.-]+\.\w{2,}$'
df = df[df['email'].str.match(email_pattern, na=False)]
```

```python
df['consent_given'] = df['consent_given'].map({
    'Yes': True, 'No': False,
    'yes': True, 'no': False,
    'TRUE': True, 'FALSE': False,
    True: True, False: False
})
```

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
```

```python
df.to_csv("survey_clean.csv", index=False)
print("Done! Cleaned data saved.")
```
